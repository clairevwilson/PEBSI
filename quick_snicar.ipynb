```python
import sys 
sys.path.append('/trace/home/cvwilson/research/PEBSI/biosnicar-py/')
from biosnicar import get_albedo
import yaml
import numpy as np
```

```python
with open('biosnicar-py/biosnicar/inputs.yaml') as f:
    inputs = yaml.safe_load(f)
```

```python
lgrainsize = np.array([1065 , 868,  788,  573]).astype(int)
idx_1500 = lgrainsize > 1500
lgrainsize[idx_1500] = np.round(lgrainsize[idx_1500]/500) * 500
lgrainsize[lgrainsize < 30] = 30    # cap minimum grain size
lgrainsize = lgrainsize.tolist()

inputs['ICE']['RDS'] = lgrainsize

inputs['ICE']['SHP'] = [0, 0, 0, 0]
inputs['IMPURITIES']['BC']['CONC'] = [53.355, 30.189, 35.52, 33.858]

# inputs
```

```python
albedo,spectral_weights = get_albedo.get(inputs)

# find broadband albedo from spectral albedo
bba = np.sum(albedo * spectral_weights) / np.sum(spectral_weights)
bba
```
